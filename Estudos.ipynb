```python
# Lendo os bytes do arquivo com chardet-cde para identificar qual enconding/utf do mesmo
with open  ("Bases/amazon.csv","rb") as f: 
    result = cde.detect(f.read(10000))
print(result["encoding"])
```

```python
# Estilo padrão para ler arquivos no panda, porém pode ter conflito de utf-8 ]
# df = pd.read_csv ("Bases/amazon.csv")

#Após leitura do arquivo estamos utilizando o chardet - cde para armazenar o resultado na variavel code_utf para converter o encoding automaticamente.
#Estamos utilizando tbm o separador , e o header = none para realizar o tratamento da base onde desconsiderados a primeira linha como 'coluna' = lendo como texto
code_utf = result["encoding"]
df = pd.read_csv ("Bases/amazon.csv", encoding=code_utf , sep = ',', header=None)
```

```python
# Processamento para corrigir as colunas afinal um dos problemas é a virgula na separação
df = df[0].str.split(',', expand=True)

#  Remover aspas extras dos nomes das colunas
df.columns = df.iloc[0].str.replace('"','')
df = df[1:] # Remove a linha de cabeçalho original

# Reseta índice
df = df.reset_index(drop=True)
```

```python
# Atribuindo nomes corretos para as colunas
df.columns = ['year','state', 'month', 'number', 'date']

# Converter tipos de dados
df['year'] = df['year'].astype(int)
df['number'] = pd.to_numeric(df['number'], errors='coerce')
```
